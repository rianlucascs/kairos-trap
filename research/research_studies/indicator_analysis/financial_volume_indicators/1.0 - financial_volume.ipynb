```python
from data_providers.providers.yfinance_price_provider import YFinancePriceProvider
```

```python
df = YFinancePriceProvider().get_asset_price(tickers="KLBN11.SA", period="max")
df = (
    df
    .assign(ret=lambda x: x["Adj Close"].pct_change(1))
    .assign(volume_ma_5=lambda x: x["Volume"].rolling(20).mean())
)

df
```

```python
import plotly.express as px
fig = px.line(
    df,
    x=df.index,
    y="volume_ma_5",
    title="KLBN11.SA — Volume",
)

fig.add_hline(df["Volume"].quantile(0.95), line_dash="dash", line_color="red")
fig.add_hline(df["Volume"].quantile(0.75), line_dash="dash", line_color="blue")
fig.add_hline(df["Volume"].quantile(0.50), line_dash="dash", line_color="green")

fig.show()
```
